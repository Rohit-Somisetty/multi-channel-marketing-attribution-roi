# Heuristic Attribution + ROI Readout
Comparing multiple rule-based attribution models to understand how channel and campaign ROI shifts under different assumptions.

## Why multiple heuristics?
- **First-touch** captures early demand capture bias and inflates upper-funnel channels.
- **Last-touch** tilts toward closing tactics (retargeting/email) and under-values awareness.
- **Linear** assumes every touch matters equally, which over-weights long journeys.
- **Time-decay** emphasizes recency via a 7-day half-life, rewarding conversion-proximate engagements.
Using all four exposes the confidence bands around ROI so finance can stress-test budget moves.

```python
from pathlib import Path

import pandas as pd
import matplotlib.pyplot as plt

plt.style.use('seaborn-v0_8')

PROJECT_ROOT = Path.cwd().resolve()
if PROJECT_ROOT.name == 'notebooks':
    PROJECT_ROOT = PROJECT_ROOT.parent

channel_df = pd.read_csv(PROJECT_ROOT / 'data' / 'attribution_channel_metrics.csv')
campaign_df = pd.read_csv(PROJECT_ROOT / 'data' / 'attribution_campaign_metrics.csv')
channel_df.head()
```

```python
pivot = channel_df.pivot(index='channel', columns='model_name', values='ROI')
ax = pivot.plot(kind='bar', figsize=(9, 5))
ax.set_ylabel('ROI ((rev-cost)/cost)')
ax.set_title('Channel ROI comparison across heuristic models')
plt.tight_layout()
plt.show()
pivot.round(2)
```

```python
top_campaigns = (
    campaign_df.sort_values(['model_name', 'attributed_revenue'], ascending=[True, False])
    .groupby('model_name')
    .head(10)
    .reset_index(drop=True)
)
top_campaigns[['model_name', 'campaign_id', 'attributed_revenue', 'ROAS', 'ROI']]
```

## Budget allocation implications
1. Organic content dominates ROI under every heuristic (ROI 215–227), suggesting incremental spend in SEO/community has room until marginal ROI collapses.
2. Paid search remains essential for volume but ROI is 6–7x, so increase only if CAC targets accept ~$65 CPA; otherwise recycle budget toward top organic/email plays.
3. Email nurture (welcome + promo) consistently clears 85x ROI with sub-$6 CPA—protect deliverability and scale triggered journeys.
4. Field and trade show events barely break even; treat them as awareness levers and tie to pipeline influence goals before approving future spend.
5. Campaign-level view shows webinar-on-demand and blog content stay top-10 regardless of heuristic, so prioritize content refresh before launching net-new programs.
